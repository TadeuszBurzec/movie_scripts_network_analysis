### Krok 1A. Wczytanie plików i raport o kolumnach

#### Prompt dla modelu

```text
Kontekst:
Masz pięć plików z poprzednich modułów analizy scenariusza filmowego.

Wejście:
Pliki: `graf_postaci.graphml`, `dialogi.csv`, `postacie.csv`, `emocje.csv`, `emocje_postaci.csv`.

Zadanie:
Wczytaj wszystkie pięć plików. Dla każdego pliku CSV wypisz: liczbę wierszy i dokładne nazwy kolumn. Dla pliku GraphML wypisz: liczbę węzłów i krawędzi. Wczytaj graf do zmiennej o nazwie `G` (networkx). Wczytaj CSV-e do zmiennych o nazwach `dialogi`, `postacie`, `emocje`, `emocje_postaci`. Sprawdź, czy kolumna z nazwą postaci istnieje we wszystkich czterech CSV-ach i pokaż kilka przykładów nazw postaci z każdego pliku.

Pokaż wynik:
- tabelkę: plik | liczba wierszy/węzłów | kolumny,

Warunek poprawności:
Nazwy postaci powinny wyglądać tak samo w obu plikach. Jeśli się różnią, zaznacz to.
```

````python
import pandas as pd
import networkx as nx
import os

files = {
    'graf_postaci.graphml': 'G',
    'dialogi.csv': 'dialogi',
    'postacie.csv': 'postacie',
    'emocje.csv': 'emocje',
    'emocje_postaci.csv': 'emocje_postaci'
}

report_data = []

try:
    # Load GraphML
    G = nx.read_graphml('graf_postaci.graphml')
    report_data.append(['graf_postaci.graphml', f"{G.number_of_nodes()} nodes, {G.number_of_edges()} edges", 'N/A'])

    # Load CSVs - postacie.csv specifically needs sep=';'
    dialogi = pd.read_csv('dialogi.csv')
    postacie = pd.read_csv('postacie.csv', sep=';')
    emocje = pd.read_csv('emocje.csv')
    emocje_postaci = pd.read_csv('emocje_postaci.csv')

    csvs = {'dialogi.csv': dialogi, 'postacie.csv': postacie, 'emocje.csv': emocje, 'emocje_postaci.csv': emocje_postaci}

    for name, df in csvs.items():
        report_data.append([name, len(df), ", ".join(df.columns)])

    # Display Table
    report_df = pd.DataFrame(report_data, columns=['Plik', 'Liczba wierszy/węzłów', 'Kolumny'])
    display(report_df)

    # Compare character names
    print("\n--- Porównanie przykładowych nazw postaci ---")

    def get_char_col(df):
        for c in df.columns:
            if any(k in c.lower() for k in ['postać', 'postac', 'character', 'name', 'speaker']):
                return c
        return None

    p_col = get_char_col(postacie)
    ep_col = get_char_col(emocje_postaci)

    if p_col and ep_col:
        # Cleaning names for comparison (stripping quotes and whitespace)
        p_sample = postacie[p_col].astype(str).str.strip('" ').unique()[:5]
        ep_sample = emocje_postaci[ep_col].astype(str).str.strip('" ').unique()[:5]

        print(f"Przykłady z 'postacie' ({p_col}):", p_sample)
        print(f"Przykłady z 'emocje_postaci' ({ep_col}):", ep_sample)

        # Simple heuristic check for intersection
        if set(p_sample).intersection(set(ep_sample)):
            print("\n✅ Znaleziono wspólne postacie w próbkach.")
        else:
            print("\n⚠️ Uwaga: Nazwy w próbkach plików 'postacie' i 'emocje_postaci' różnią się.")
    else:
        print("❌ Nie udało się automatycznie zidentyfikować kolumny postaci.")

except Exception as e:
    print(f"⛔ Błąd podczas wczytywania: {e}")
````

````python
# NORMALIZACJA — ujednolicenie nazwy kolumny postaci we wszystkich tabelach
# Uruchom zaraz po komórce z wczytaniem plików. Działa automatycznie.

_SZUKAJ = ('postac', 'postać', 'character', 'speaker', 'bohater',
           'imie', 'imię', 'name', 'nazwa', 'person', 'char')
_KANON  = 'postać'

_tabele = {
    'dialogi':        dialogi,
    'postacie':       postacie,
    'emocje':         emocje,
    'emocje_postaci': emocje_postaci,
}

print("🔍 Wykrywanie i normalizacja kolumny postaci:\n")
_problem = False
for nazwa, df in _tabele.items():
    kol = next((c for c in df.columns
                if any(k in c.lower() for k in _SZUKAJ)), None)
    if kol is None:
        print(f"  ❌ {nazwa}: nie znaleziono — dostępne kolumny: {list(df.columns)}")
        _problem = True
    elif kol != _KANON:
        df.rename(columns={kol: _KANON}, inplace=True)
        print(f"  🔄 {nazwa}: '{kol}' → '{_KANON}'")
    else:
        print(f"  ✅ {nazwa}: '{_KANON}' — OK")

if _problem:
    raise ValueError(
        "\n⛔ Nie udało się automatycznie znaleźć kolumny postaci w jednym z plików.\n"
        "Sprawdź listę kolumn wypisaną powyżej i dodaj poniżej ręczne przemianowanie:\n\n"
        "    dialogi.rename(columns={'TWOJA_NAZWA': 'postać'}, inplace=True)\n\n"
        "Następnie uruchom tę komórkę ponownie."
    )
print(f"\n✅ Wszystkie tabele mają kolumnę '{_KANON}'. Przejdź do Etapu 2.")
````

### Krok 2A. Wzbogacenie grafu i zapis bundle.json

#### Cel i sens analityczny

Wzbogacony graf łączy wszystko: wiesz nie tylko, kto z kim rozmawia, ale czyim głosem dominuje scena, jaką emocją jest nacechowany i jakiej jest płci. `bundle.json` zbiera te dane w formacie gotowym do wklejenia do modelu generującego prezentację.

#### Prompt dla modelu

```text
Kontekst:
Masz wczytane: graf sieci w zmiennej `G` (networkx), tabele `postacie`, `emocje_postaci` i `emocje`. Nazwy kolumn mogą się różnić od podanych — użyj tych, które faktycznie są w plikach.

Wejście:
Graf `G` i cztery tabele wczytane w Etapie 1.

Zadanie:
Wykonaj trzy rzeczy:

1. WZBOGAĆ GRAF: dla każdego węzła grafu dołącz atrybuty z `postacie` i `emocje_postaci`: płeć (gender), liczba słów (word_count), dominująca emocja (dominant_emotion), średnia walencja (avg_valence). Węzły bez danych: gender="?", word_count=0, dominant_emotion="neutralna", avg_valence=0.0. Zapisz wzbogacony graf do `graf_postaci_wzbogacony.graphml`.

2. OBLICZ AGREGATY:
   a. top_speakers: top 15 postaci wg liczby słów — z płcią i procentem łącznej mowy.
   b. speech_by_gender: łączna mowa wg K/M/? — i ich procent.
   c. emotion_arc: dla każdej sceny: średnia walencja, wygładzona walencją (okno 5). Lista par (numer_sceny, rolling_valence).
   d. emotion_by_gender: dla K i M — ile % kwestii przypada na każdą z 6 emocji.

3. ZAPISZ bundle.json (UTF-8, indent=2):
{
  "top_speakers": [{"name":..., "gender":..., "word_count":..., "pct_of_total":...}],
  "speech_by_gender": {"K":{"total_words":..., "pct":...}, "M":{...}, "?":{...}},
  "emotion_arc": [{"scene":..., "avg_valence":..., "rolling_valence":...}],
  "emotion_by_gender": {
    "K": {"radość":..., "smutek":..., "złość":..., "strach":..., "zaskoczenie":..., "neutralna":...},
    "M": {tak samo}
  },
  "network": {
    "nodes": [{"id":..., "gender":..., "word_count":..., "scene_count":..., "dominant_emotion":..., "avg_valence":...}],
    "edges": [{"source":..., "target":..., "weight":...}]
  }
}
Zachowaj dane bundle jako zmienną o nazwie `bundle`.

Pokaż wynik:
- potwierdzenie zapisu obu plików z rozmiarami,
- 5 przykładowych węzłów z wzbogaconego grafu (postać + atrybuty),
- pierwsze i ostatnie 3 wpisy łuku emocjonalnego.

Warunek poprawności:
Suma % speech_by_gender = 100%. Procenty emocji dla K sumują się do 100%, tak samo dla M.

Jeśli wystąpi błąd:
Wypisz, na którym kroku (wzbogacanie / agregaty / bundle) wystąpił problem.
```

````python
import json
import pandas as pd
import networkx as nx
import os

try:
    # 1. WZBOGAĆ GRAF
    # Merging character data
    char_data = pd.merge(postacie, emocje_postaci, on='postać', how='outer')

    for node in G.nodes():
        row = char_data[char_data['postać'] == node]
        if not row.empty:
            # Safely handle potential NaNs using pd.isna check or fillna logic
            gender = row.iloc[0].get('plec', '?')
            words = row.iloc[0].get('liczba_slow', 0)
            emotion = row.iloc[0].get('dominująca_emocja', 'neutralna')
            valence = row.iloc[0].get('średnia_walencja', 0.0)

            G.nodes[node]['gender'] = str(gender) if pd.notna(gender) else '?'
            G.nodes[node]['word_count'] = int(words) if pd.notna(words) else 0
            G.nodes[node]['dominant_emotion'] = str(emotion) if pd.notna(emotion) else 'neutralna'
            G.nodes[node]['avg_valence'] = float(valence) if pd.notna(valence) else 0.0
        else:
            G.nodes[node]['gender'] = '?'
            G.nodes[node]['word_count'] = 0
            G.nodes[node]['dominant_emotion'] = 'neutralna'
            G.nodes[node]['avg_valence'] = 0.0

    nx.write_graphml(G, 'graf_postaci_wzbogacony.graphml')

    # 2. OBLICZ AGREGATY
    # a. top_speakers
    total_words = postacie['liczba_slow'].sum()
    top_speakers_df = postacie.sort_values('liczba_slow', ascending=False).head(15)
    top_speakers = [
        {
            "name": str(row['postać']),
            "gender": str(row['plec']),
            "word_count": int(row['liczba_slow']),
            "pct_of_total": round((row['liczba_slow'] / total_words) * 100, 2)
        } for _, row in top_speakers_df.iterrows()
    ]

    # b. speech_by_gender
    sbg = postacie.groupby('plec')['liczba_slow'].sum()
    total_sbg = sbg.sum()
    speech_by_gender = {
        str(g): {"total_words": int(sbg[g]), "pct": round((sbg[g] / total_sbg) * 100, 2)}
        for g in sbg.index
    }

    # c. emotion_arc
    scene_emotions = emocje.groupby('numer_sceny')['walencja'].mean().reset_index()
    scene_emotions['rolling_valence'] = scene_emotions['walencja'].rolling(window=5, min_periods=1, center=True).mean()
    emotion_arc = [
        {"scene": int(row['numer_sceny']), "avg_valence": round(float(row['walencja']), 3), "rolling_valence": round(float(row['rolling_valence']), 3)}
        for _, row in scene_emotions.iterrows()
    ]

    # d. emotion_by_gender
    emocje_gender = pd.merge(emocje, postacie[['postać', 'plec']], on='postać', how='left')
    ebg_raw = emocje_gender.groupby(['plec', 'emocja']).size().unstack(fill_value=0)
    emotion_by_gender = {}
    for gender in ['K', 'M']:
        if gender in ebg_raw.index:
            row_sum = ebg_raw.loc[gender].sum()
            emotion_by_gender[gender] = (ebg_raw.loc[gender] / row_sum * 100).round(2).to_dict()
        else:
            # Handle missing gender categories
            emotion_by_gender[gender] = {e: 0.0 for e in emocje['emocja'].unique()}

    # 3. ZAPISZ bundle.json
    network_data = {
        "nodes": [{"id": n, **G.nodes[n]} for n in G.nodes()],
        "edges": [{"source": u, "target": v, "weight": float(d.get('weight', 1))} for u, v, d in G.edges(data=True)]
    }

    bundle = {
        "top_speakers": top_speakers,
        "speech_by_gender": speech_by_gender,
        "emotion_arc": emotion_arc,
        "emotion_by_gender": emotion_by_gender,
        "network": network_data
    }

    with open('bundle.json', 'w', encoding='utf-8') as f:
        json.dump(bundle, f, indent=2, ensure_ascii=False)

    # Final confirmation output
    print(f"✅ Zapisano: graf_postaci_wzbogacony.graphml ({os.path.getsize('graf_postaci_wzbogacony.graphml')/1024:.1f} KB)")
    print(f"✅ Zapisano: bundle.json ({os.path.getsize('bundle.json')/1024:.1f} KB)")

    print("\n--- 5 przykładowych węzłów ---")
    for node in list(G.nodes())[:5]:
        print(f"{node}: {G.nodes[node]}")

    print("\n--- Łuk emocjonalny (3 pierwsze i 3 ostatnie) ---")
    print(emotion_arc[:3])
    print(emotion_arc[-3:])

except Exception as e:
    print(f"⛔ Błąd na kroku wzbogacania/agregacji: {e}")
    import traceback
    traceback.print_exc()
````

````python
# TWOJA KOLEJ: uzupełnij jedną lukę [UZUPEŁNIJ: …], potem uruchom.
moj_prompt = """
Kontekst:
Masz wzbogacony graf sieci postaci w zmiennej `G` (networkx). Węzły mają atrybuty: `gender` (K/M/?), `word_count` (liczba słów), `dominant_emotion`, `avg_valence`. Krawędzie mają atrybut `weight` (liczba wspólnych scen).

Wejście:
Zmienna `G` z Etapu 2. Biblioteki `networkx` i `matplotlib` są już zainstalowane.

Zadanie:
1. Pobierz `word_count` dla każdego węzła i przeskaluj na rozmiary punktów: minimum 300, maksimum 3000 (dla węzłów z word_count=0 użyj 100).
2. Przypisz kolory wg `gender`: K = '#e07b8a', M = '#5b8db8', ? = '#aaaaaa'.
3. Ustal grubość krawędzi proporcjonalnie do `weight`: minimum 0.3, maksimum 3.0.
4. Wybierz 8 postaci z największym 'word_count' tylko dla nich wyświetl etykiety z imionami.
5. Narysuj sieć z układem spring_layout (seed=42). Dodaj tytuł "Sieć postaci — głos i płeć" i legendę kolorów płci.
6. Zapisz jako `siec_wzbogacona.png` (dpi=150).

Pokaż wynik:
Wyświetl rysunek w notatniku i potwierdź zapis pliku z jego rozmiarem.

Warunek poprawności:
Węzły mają różne rozmiary (nie wszystkie jednakowe). Plik `siec_wzbogacona.png` istnieje na dysku.

Jeśli wystąpi błąd:
Wypisz nazwę atrybutu, który sprawia problem. Sprawdź, czy `list(G.nodes(data=True))[:3]` zawiera oczekiwane klucze.

Nie rób jeszcze:
Nie modyfikuj grafu ani nie zapisuj żadnych innych plików.
"""
assert "[UZUPEŁNIJ" not in moj_prompt, \
    "⛔ Pozostała luka [UZUPEŁNIJ: …] — wpisz liczbę postaci z etykietami."
assert "siec_wzbogacona.png" in moj_prompt, \
    "⛔ Prompt musi zawierać nazwę pliku wynikowego: siec_wzbogacona.png."
print("✅ Prompt gotowy. Skopiuj poniższą treść do asystenta AI:\n")
print(moj_prompt)
````

````python
import matplotlib.pyplot as plt
import networkx as nx
import os

try:
    # 1. Prepare node sizes based on word_count
    word_counts = [d.get('word_count', 0) for n, d in G.nodes(data=True)]
    min_wc, max_wc = min(word_counts), max(word_counts)

    node_sizes = []
    for wc in word_counts:
        if wc == 0:
            node_sizes.append(100)
        else:
            # Scale wc to range [300, 3000]
            size = 300 + (wc - min_wc) * (3000 - 300) / (max_wc - min_wc if max_wc > min_wc else 1)
            node_sizes.append(size)

    # 2. Prepare node colors based on gender
    color_map = {'K': '#e07b8a', 'M': '#5b8db8', '?': '#aaaaaa'}
    node_colors = [color_map.get(d.get('gender'), '#aaaaaa') for n, d in G.nodes(data=True)]

    # 3. Prepare edge weights for thickness
    weights = [d.get('weight', 1) for u, v, d in G.edges(data=True)]
    min_w, max_w = min(weights), max(weights)
    edge_widths = [0.3 + (w - min_w) * (3.0 - 0.3) / (max_w - min_w if max_w > min_w else 1) for w in weights]

    # 4. Labels for top 8 speakers
    top_8_nodes = sorted(G.nodes(data=True), key=lambda x: x[1].get('word_count', 0), reverse=True)[:8]
    labels = {node: node for node, data in top_8_nodes}

    # 5. Drawing
    plt.figure(figsize=(12, 10))
    pos = nx.spring_layout(G, seed=42)

    nx.draw_networkx_edges(G, pos, width=edge_widths, alpha=0.3, edge_color='gray')
    nx.draw_networkx_nodes(G, pos, node_size=node_sizes, node_color=node_colors, alpha=0.9)
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_weight='bold')

    # Legend
    from matplotlib.lines import Line2D
    legend_elements = [Line2D([0], [0], marker='o', color='w', label='Kobieta (K)', markerfacecolor='#e07b8a', markersize=10),
                       Line2D([0], [0], marker='o', color='w', label='Mężczyzna (M)', markerfacecolor='#5b8db8', markersize=10),
                       Line2D([0], [0], marker='o', color='w', label='Nieznana (?)', markerfacecolor='#aaaaaa', markersize=10)]
    plt.legend(handles=legend_elements, loc='upper left', title="Płeć")

    plt.title("Sieć postaci — głos i płeć", fontsize=15)
    plt.axis('off')

    # 6. Save and Show
    plt.savefig('siec_wzbogacona.png', dpi=150, bbox_inches='tight')
    plt.show()

    print(f"✅ Zapisano plik: siec_wzbogacona.png ({os.path.getsize('siec_wzbogacona.png')/1024:.1f} KB)")

except Exception as e:
    print(f"⛔ Błąd: {e}")
    # Debug info
    print("Przykładowe dane węzłów:", list(G.nodes(data=True))[:3])
````
